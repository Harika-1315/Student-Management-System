###Question 1: Correcting a Malformed JSON String

The following string is intended to be valid JSON but contains several errors. Identify all errors, correct them, and parse the JSON into a Python dictionary. Print the resulting dictionary.

````python
{
  'name': 'Alice',
  age: 30,
  "city": "New York",
  "hobbies": ["reading", "hiking",]
}
````

````python
import json

malformed_json = """{
  'name': 'Alice',
  age: 30,
  "city": "New York",
  "hobbies": ["reading", "hiking",]
}"""

import json

malformed_json = """{
    'name': 'Alice',
    'age': 30,
    "city": "New York",
    "hobbies": ["reading", "hiking",]
}"""

# Correct the malformed JSON
corrected_json = """{
    "name": "Alice",
    "age": 30,
    "city": "New York",
    "hobbies": ["reading", "hiking"]
}"""

# Parse the corrected JSON into a Python dictionary
data = json.loads(corrected_json)

# Print the resulting dictionary
print(data)
````

The JSON was invalid because JSON requires double quotes (") around keys and string values, but single quotes (') were used for some of them. There was also a trailing comma after "hiking" in the hobbies list, which is not allowed in standard JSON. I corrected these errors and then used json.loads() to parse the corrected JSON string into a Python dictionary

#Question 2: Formatting a Python Dictionary to JSON

Convert the given Python dictionary to a JSON string with the following requirements:
- Indentation of 4 spaces
- Keys sorted alphabetically
- Non-ASCII characters preserved (ensure_ascii=False)
- Print the resulting JSON string.

````python
import json

data = {"name": "José", "age": 30, "city": "München", "hobbies": ["reading", "hiking"]}

# Convert the dictionary to a formatted JSON string
json_string = json.dumps(
    data,
    indent=4,
    sort_keys=True,
    ensure_ascii=False
)

# Print the resulting JSON string
print(json_string)
````

I used json.dumps() to convert the Python dictionary into a JSON string. I used indent=4 for formatting, sort_keys=True to sort the keys alphabetically, and ensure_ascii=False to preserve non-ASCII characters such as é and ü.



## Question 3: Extracting and Formatting Nested JSON
Given the following JSON string representing a list of users:
1.   Extract the email of the user with id=2 and print it.
2.   Pretty-print the entire JSON with indentation of 2 spaces.



````python
import json

json_str = '''
[
  {"id": 1, "name": "Alice", "email": "alice@example.com"},
  {"id": 2, "name": "Bob", "email": "bob@example.com"},
  {"id": 3, "name": "Charlie", "email": "charlie@example.com"}
]
'''

# Parse the JSON string into a Python object
users = json.loads(json_str)

# Find and print the email of the user with id = 2
for user in users:
    if user["id"] == 2:
        print("Email:", user["email"])
        break

# Pretty-print the entire JSON with 2-space indentation
print(json.dumps(users, indent=2))
````

I used json.loads() to convert the JSON string into a Python list of dictionaries. I searched for the user with id equal to 2 and printed the email address. Finally, I used json.dumps() with indent=2 to pretty-print the complete JSON data.



## Question 4: Handling Duplicate Keys

The following JSON string contains duplicate keys:

```
{"a": 1, "b": 2, "a": 3}
```


1.   Explain what happens when you parse it with json.loads.
2.   Write code to parse it and merge duplicate keys by summing their values. Print the result.




````python
import json

dup_json = '{"a": 1, "b": 2, "a": 3}'

# Custom function to merge duplicate keys by summing their values
def merge_duplicate_keys(pairs):
    result = {}

    for key, value in pairs:
        if key in result:
            result[key] += value
        else:
            result[key] = value

    return result


# Parse the JSON and handle duplicate keys
result = json.loads(dup_json, object_pairs_hook=merge_duplicate_keys)

# Print the merged result
print(result)
````

Explanation: What does Python's json module do with duplicate keys? How did you merge them?

Python's json.loads() normally keeps only the last value when duplicate keys occur. I used object_pairs_hook to process all key-value pairs and merge duplicate keys by summing their values. Therefore, the two values for "a" (1 and 3) were combined to produce "a": 4.



Question 5: Fixing JSON with Comments and Trailing Commas

The following string is not valid JSON because it contains // comments and a trailing comma. Remove the comments and the trailing comma, then parse it into a Python dictionary and print it.


```
{
  // user name
  "name": "Alice",
  "age": 30, // age in years
  "city": "New York",
}
```



````python
import json
import re

invalid_json = '''
{
  // user name
  "name": "Alice",
  "age": 30, // age in years
  "city": "New York",
}
'''

# Remove // comments from the JSON string
cleaned_json = re.sub(r'//.*', '', invalid_json)

# Remove trailing commas before } or ]
cleaned_json = re.sub(r',\s*([}\]])', r'\1', cleaned_json)

# Parse the cleaned JSON
data = json.loads(cleaned_json)

# Print the resulting dictionary
print(data)
````

Explanation: Describe the steps you took to clean the string before parsing.

The JSON was invalid because it contained // comments and a trailing comma. I first used a regular expression to remove the comments, then removed the trailing comma before the closing brace. Finally, I used json.loads() to parse the cleaned JSON string into a Python dictionary.



Question 6: Validating and Formatting a JSON String

Write a function validate_and_format(json_string) that:
- Attempts to parse the given JSON string.
- If parsing fails, returns "Invalid JSON".
- If parsing succeeds, returns the JSON string pretty-printed with indentation of 4 spaces and sorted keys.
- Test it with both a valid and an invalid JSON string.

````python
import json

def validate_and_format(json_string):
    try:
        # Parse the JSON string
        data = json.loads(json_string)

        # Return formatted JSON with 4-space indentation and sorted keys
        return json.dumps(data, indent=4, sort_keys=True)

    except json.JSONDecodeError:
        return "Invalid JSON"


# Test cases
valid = '{"b": 2, "a": 1}'
invalid = '{"a": 1, "b": 2,}'

print(validate_and_format(valid))
print(validate_and_format(invalid))
````



The function uses json.loads() to validate the JSON string. If parsing fails, it returns "Invalid JSON". If parsing succeeds, it uses json.dumps() with 4-space indentation and sorted keys to format the JSON string.
